# Partie 1 – Graphique linéaire (Line Plot) 

1) Afficher l'évolution de la température en fonction du temps, avec le titre, les labels, la légende 
et la grille. 

```python
plt.figure(figsize=(12, 6))

plt.plot(df["date_heure"], df["temperature"], label="Température")

plt.title("Évolution de la température en fonction du temps")
plt.xlabel("Temps")
plt.ylabel("Température (°C)")

plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

2) Identifier visuellement les valeurs qui semblent anormales. 

Les temperatures qui semblent anormales avoisinent les -20 degres (temperatures faibles) et les 60 degres (temperatures elevees)

# Partie 2 – Diagramme en barres (Bar Chart) 

1) Déterminer consommation_batiment qui stocke les consommations moyennes des bâtiments 

```python
consommation_batiment = df.groupby("batiment")["consommation"].mean()
consommation_batiment
```

2) Comparer graphiquement la consommation moyenne des bâtiments

```python
plt.figure(figsize=(10, 6))
consommation_batiment.plot(kind="bar")
plt.title("Consommation moyenne par bâtiment")
plt.xlabel("Bâtiment")
plt.ylabel("Consommation moyenne")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

3) Comparer la consommation moyenne des bâtiments avec un graphique horizontal 

```python
plt.figure(figsize=(10, 6))
consommation_batiment.plot(kind="barh")
plt.title("Consommation moyenne par bâtiment")
plt.xlabel("Consommation moyenne")
plt.ylabel("Bâtiment")
plt.grid(axis="x")
plt.tight_layout()
plt.show()
```

4) Dans quels cas un graphique horizontal peut-il être plus lisible ? 

Le graphique horizontal est particulièrement lisible lorsque les catégories possèdent des noms longs ou lorsqu'il existe un grand nombre de catégories. Il facilite également la comparaison entre les différentes valeurs.

# Partie 3 – Histogramme 

1) Afficher la distribution des températures en divisant la plage des données en 20 classes et en 
affichant uniquement les lignes horizontales de la grille

```python
plt.figure(figsize=(10, 6))

plt.hist(df["temperature"].dropna(), bins=20)

plt.title("Distribution des températures")
plt.xlabel("Température (°C)")
plt.ylabel("Fréquence")

plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

2) Analyser le graphique pour voir 

a) -  où se concentrent les températures ; 

Les temperatures les plus eleves sont concentrees dans l'intervalle de temperatures  [20 , 30] 

b) si la distribution semble symétrique ;  

La distibution ne semble pas symetrique car les vaaleurs de gauches et de droite ne sont pas repartis de maniere similaire par rapport au centre.

c) s'il existe des valeurs très éloignées ; 

Oui il en existe. Car on retrouve des temperatures à -20 degres et à 60 degres.

d) si des valeurs aberrantes sont visibles. 

Oui il existe des valeurs valeurs tres eloignees de la zone de distribution donc nous pouvons conclure que ce sont des valeurs abberrantes.

3) Afficher la distribution de la consommation en divisant la plage des données en 20 classes

```python
plt.figure(figsize=(10, 6))

plt.hist(df["consommation"].dropna(), bins=20)

plt.title("Distribution de la consommation")
plt.xlabel("Consommation")
plt.ylabel("Fréquence")

plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

a) Modifier le nombre de classes à 10 

```python
plt.figure(figsize=(10, 6))

plt.hist(df["consommation"].dropna(), bins=10)

plt.title("Distribution de la consommation - 10 classes")
plt.xlabel("Consommation")
plt.ylabel("Fréquence")

plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

b) Modifier le nombre de classes à 30 

```python
plt.figure(figsize=(10, 6))

plt.hist(df["consommation"].dropna(), bins=30)

plt.title("Distribution de la consommation - 30 classes")
plt.xlabel("Consommation")
plt.ylabel("Fréquence")

plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

c) Quel impact le nombre de classes a-t-il sur l'interprétation ? 

Le nombre de classes influence fortement la lecture de l'histogramme. Avec peu de classes, la distribution est plus générale et certaines caractéristiques peuvent être masquées. Avec beaucoup de classes, le graphique devient plus détaillé mais peut également paraître plus irrégulier. Il faut donc choisir un nombre de classes permettant de trouver un équilibre entre simplicité et précision.

# Partie 4 – Nuage de points (Scatter Plot) 

1) Réaliser un nuage de points entre température et consommation 

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["temperature"],
    df["consommation"],
    alpha=0.6
)

plt.title("Relation entre température et consommation")
plt.xlabel("Température (°C)")
plt.ylabel("Consommation")

plt.grid(True)

plt.tight_layout()
plt.show()
```

2) Déterminer visuellement s'il semble exister une relation entre les deux 

Dans ce nuage de point il apparait qu'il y a une relation faible entre les temperatures et les consommations. 

# Partie 5 – Diagramme à moustache (Box plot)

1) Créer un box plot de la température. N’oublier pas de supprimer les valeurs manquantes avant 
de construire le graphique et afficher uniquement les lignes horizontales de la grille 

```python
temperature = df["temperature"].dropna()
```

```python
plt.figure(figsize=(8, 6))

plt.boxplot(temperature)

plt.title("Box plot de la température")
plt.ylabel("Température (°C)")

plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

2) Que représentent les points situés en dehors des moustaches ? 

Les points situés en dehors des moustaches représentent généralement des valeurs atypiques ou potentiellement aberrantes. Elles sont éloignées de la majorité des observations selon la règle utilisée pour construire le box plot.

3) Créer un box plot de la consommation 

```python
consommation = df["consommation"].dropna()

plt.figure(figsize=(8, 6))

plt.boxplot(consommation)

plt.title("Box plot de la consommation")
plt.ylabel("Consommation")

plt.grid(axis="y")

plt.tight_layout()
plt.show()
```

Nous remarquons que la mediane du Boxplot des temperatures est situee entre 20 et 30 degres.
Pour le Boxplot des consommations, la medianne est situee juste au dessus de 200.

b) la taille de la boîte ;  

Le Boxplox des temperatures est plus applatit et celui des consommations est beaucoup plus grande indiquant une forte dispersion des donnees.

c) la longueur des moustaches ;  

La longueur des moustaches est plus longues dans le Boxplot des consommations montrant l'étendue des valeurs non considérées comme atypiques.

d) les valeurs extrêmes. 

Il existe effectivement des valeurs aberrantes representes par les points isoles.

# Partie 6 – Diagramme circulaire (Pie Chart) 

1) Réaliser un diagramme pour analyser la répartition des états (OK, ALERTE, ERREUR) 

```python
repartition_etats = df["etat"].value_counts()

repartition_etats
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    repartition_etats,
    labels=repartition_etats.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Répartition des états des capteurs")

plt.show()
```

2) interpréter la proportion de capteurs OK ; proportion ALERTE ;  proportion ERREUR 

```python
proportions = df["etat"].value_counts(normalize=True) * 100

proportions
```

-     Interpretation


Les capteurs OK representent la proportion de capteurs fonctionnant  normalement est occupent 94,34%.
ALERTE : la proportion de capteurs nécessitant une surveillance avec un taux de 4.82%
ERREUR :la  proportion de capteurs présentant une erreur qvec un taux de 0.83%.


# Partie 7 : Plusieurs courbes sur un même graphique

Construire sur un même graphique  « Évolution de la température par bâtiment" » des quatre 
bâtiments.

```python
plt.figure(figsize=(12, 6))

for batiment in df["batiment"].unique():
    donnees = df[df["batiment"] == batiment]
    
    plt.plot(
        donnees["date_heure"],
        donnees["temperature"],
        label=batiment
    )

plt.title("Évolution de la température par bâtiment")
plt.xlabel("Temps")
plt.ylabel("Température (°C)")

plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
